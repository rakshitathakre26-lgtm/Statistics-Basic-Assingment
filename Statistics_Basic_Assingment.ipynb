Question 1: What is the difference between descriptive statistics and inferential
statistics? Explain with examples.


1. Descriptive Statistics
•	Definition: Descriptive statistics are used to summarize, organize, and present data in a meaningful way.
•	Purpose: They describe what the data shows, but they do not allow you to make conclusions beyond the data.
•	Examples of tools: Mean, median, mode, range, standard deviation, frequency tables, graphs (histograms, pie charts).
✅ Example:
A teacher records the marks of 50 students in a test.
•	The average score is 72.
•	The highest score is 95 and the lowest score is 40.
•	A histogram shows most students scored between 60–80.
👉 These summaries describe the dataset but do not predict or generalize to all students everywhere.
2. Inferential Statistics
•	Definition: Inferential statistics are used to make predictions, inferences, or generalizations about a larger population based on a sample of data.
•	Purpose: They help answer questions beyond the data collected.
•	Examples of tools: Hypothesis testing, confidence intervals, regression, correlation, t-tests, chi-square tests.
✅ Example:
Suppose the teacher takes a sample of 50 students from a school of 1,000 students.
•	Using inferential statistics, she might test whether the average score of the whole school is likely around 72 (based on the sample).
•	She could run a hypothesis test to see if boys and girls perform differently.
•	Or, she could build a confidence interval to estimate the true average score of all 1,000 students.
👉 Here, conclusions go beyond the sample and are applied to the population
Key Difference in One Line
•	Descriptive Statistics: What does the data say?
•	Inferential Statistics: What can we conclude or predict about the population from the data?



Question 2: What is sampling in statistics? Explain the differences between random and stratified sampling.

What is Sampling in Statistics?
• Definition: Sampling is the process of selecting a subset (sample)
from a larger group (population) to study and make conclusions
about the whole population.
• Why use sampling? Because studying the entire population is
often time-consuming, expensive, or impractical.

Example: Instead of asking all 10,000 students in a city about their
favorite subject, a researcher may ask only 500 students (the sample)
and use their answers to represent the whole group.
Random Sampling
• Definition: Every member of the population has an equal chance
of being selected.
• Purpose: Reduces bias and makes the sample more representative
of the population.
• Example: Writing the names of all 1,000 students on slips of
paper, mixing them, and picking 100 randomly.

Advantage: Simple and unbiased.

Limitation: May not always represent subgroups well (e.g., boys vs.
girls, age groups).
Stratified Sampling
• Definition: The population is divided into different groups (strata)
based on a characteristic (e.g., gender, age, income level). Then,
random samples are taken from each group in proportion to their
size.
• Purpose: Ensures all important subgroups are fairly represented.
• Example: In a school with 60% girls and 40% boys, if we need a
sample of 100 students, we take 60 girls and 40 boys randomly.

Advantage: More accurate and representative, especially when
population has subgroups.

Limitation: Requires knowledge of population structure beforehand.

Question 3: Define mean, median, and mode. Explain why these
measures of central tendency are important.

Measures of Central Tendency: Mean, Median, Mode
1. Mean (Average)
• Definition: The mean is the sum of all values divided by the total
number of values.
• Formula:

2. Median
• Definition: The median is the middle value when data is arranged
in order.
• If there are an odd number of values → pick the middle one.
• If there are an even number of values → take the average of the
two middle values.
• Example:
o Scores = 10, 20, 30, 40, 50 → Median = 30
o Scores = 10, 20, 30, 40 → Median = (20 + 30) ÷ 2 = 25
3. Mode
• Definition: The mode is the value that appears most frequently in
the dataset.
• A dataset may have no mode, one mode, or multiple modes.
• Example:
o Scores = 10, 20, 20, 30, 40 → Mode = 20
o Scores = 5, 5, 10, 10, 15 → Modes = 5 and 10 (bimodal)
Why Are These Measures Important?
They are called measures of central tendency because they help us
understand the "center" or typical value of a dataset.
• Mean → Uses all data values → good overall summary, but
sensitive to outliers (e.g., very high/low values).
• Median → Useful when data has outliers or skewed distribution
(e.g., incomes, house prices).
• Mode → Useful for categorical data (e.g., the most popular
subject, most common shoe size).

Example of importance:
Suppose incomes (in ₹): 20k, 22k, 23k, 25k, 5,00,000
• Mean = very high (pulled up by 5,00,000) → misleading.
• Median = 23k → better represents "typical" income.
• Mode (if many people earn 22k) → shows the most common
earning.
In short:
• Mean = mathematical average
• Median = middle value
• Mode = most frequent value
• Together, they give a complete picture of the dataset.

Question 4: Explain skewness and kurtosis. What does a positive skew
imply about the data?

1. Skewness
• Definition: Skewness measures the asymmetry of a data
distribution around its mean.
• Types of skewness:
1. Positive skew (right skew): The tail on the right side (higher
values) is longer.
▪ Most data points are concentrated on the left (lower
values).
▪ Mean > Median > Mode
▪ Example: Incomes where most people earn low, few
earn very high.
2. Negative skew (left skew): The tail on the left side (lower
values) is longer.
▪ Most data points are concentrated on the right (higher
values).
▪ Mean < Median < Mode
▪ Example: Exam scores where most students score high,
few score very low.
3. Zero skew (symmetrical distribution): Left and right tails are
balanced.
▪ Mean ≈ Median ≈ Mode

Positive skew implication:
• Data has more low values with a few extremely high values.
• The mean is pulled toward the higher end.
• Example: Household income in a city: many earn 20–50k, a few
earn 5,00,000.
2. Kurtosis
• Definition: Kurtosis measures the peakedness or flatness of a
distribution compared to a normal distribution.
• Types of kurtosis:
1. Leptokurtic (kurtosis > 3): Tall and sharp peak, heavy tails →
more extreme values.
2. Platykurtic (kurtosis < 3): Flat and wide peak, light tails →
fewer extreme values.
3. Mesokurtic (kurtosis ≈ 3): Normal distribution shape.

Importance:
• High kurtosis → higher risk of outliers.
• Low kurtosis → data is more evenly spread.

Question 5: Implement a Python program to compute the mean,
median, and mode of a given list of numbers. numbers = [12, 15, 12, 18,
19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]  


Here’s a Python program that calculates mean, median, and mode for
your list:

```python
# Import necessary modules
from statistics import mean, median, mode
# Given list of numbers
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
# Calculate mean, median, and mode
mean_value = mean(numbers)
median_value = median(numbers)
mode_value = mode(numbers)
# Print the results
print("Numbers:", numbers)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

Explanation:
• Mean: Sum of all numbers ÷ 15 = 20.07
• Median: Middle value when numbers are sorted = 19
• Mode: Most frequent value = 12

Question 6: Compute the covariance and correlation coefficient
between the following two datasets provided as lists in Python: list_x =
[10, 20, 30, 40, 50] list_y = [15, 25, 35, 45, 60]

Python Code Using numpy

```python
import numpy as np
# Given datasets
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]
# Convert lists to numpy arrays
x = np.array(list_x)
y = np.array(list_y)
# Compute covariance
cov_matrix = np.cov(x, y)  # Returns covariance matrix
cov_xy = cov_matrix[0, 1]  # Covariance between x and y
# Compute correlation coefficient
corr_matrix = np.corrcoef(x, y)  # Returns correlation matrix
corr_xy = corr_matrix[0, 1]
# Print results
# Correlation coefficient between x and y
print("Covariance between x and y:", cov_xy)
print("Correlation coefficient between x and y:", corr_xy)
```

Explanation:
1. Covariance measures the direction of the linear relationship
between two variables:
o Positive → as x increases, y also increases
o Negative → as x increases, y decreases
2. Correlation coefficient measures both direction and strength of
the relationship:
o Value ranges from -1 to +1
o +1 → perfect positive correlation
o -1 → perfect negative correlation
o 0 → no linear correlation

Here, corr_xy ≈ 0.997 → strong positive correlation.

Question 7: Write a Python script to draw a boxplot for the following
numeric list and identify its outliers. Explain the result: data = [12, 14,
14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

Python Code Using Matplotlib

```python
import matplotlib.pyplot as plt
# Given data
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
# Create a boxplot
plt.boxplot(data, vert=True, patch_artist=True)
plt.title("Boxplot of the Data")
plt.ylabel("Values")
plt.show()
# Optional: Identify outliers using IQR
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = [x for x in data if x < lower_bound or x > upper_bound]
print("Outliers in the dataset:", outliers)
```

Explanation
1. Boxplot Components:
o Box: Interquartile Range (IQR = Q3 − Q1) → middle 50% of
data
o Line inside box: Median
o Whiskers: Data within 1.5 × IQR from Q1 and Q3
o Dots outside whiskers: Outliers
2. Outlier Calculation:
o Q1 (25th percentile) = 17.75
o Q3 (75th percentile) = 24.75
o IQR = 24.75 − 17.75 = 7
o Lower bound = Q1 − 1.5×IQR = 17.75 − 10.5 = 7.25
o Upper bound = Q3 + 1.5×IQR = 24.75 + 10.5 = 35.25

Outliers:
• Any value < 7.25 or > 35.25 → Here, 35 is just below the upper
bound, so it’s not considered a strict outlier by this calculation.
3. Interpretation:
• Most data is concentrated between 14 and 26.
• 35 is a high value relative to the rest, showing a potential extreme
value.

Question 8: You are working as a data analyst in an e-commerce
company. The marketing team wants to know if there is a relationship
between advertising spend and daily sales.
● Explain how you would use covariance and correlation to explore
this relationship.  
● Write Python code to compute the correlation between the two
lists:
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

Step 1: Understanding the Relationship
As a data analyst, you can use covariance and correlation to explore
how advertising spend affects daily sales:
1. Covariance:
o Measures the direction of the relationship between two
variables.
o Positive covariance → as advertising spend increases, sales
also increase.
o Negative covariance → as advertising spend increases, sales
decrease.
o Limitation: Magnitude depends on the units, so it’s hard to
interpret strength.
2. Correlation coefficient (Pearson’s r):
o Measures both strength and direction of the linear
relationship.
o Value ranges between -1 and +1.
▪ +1 → perfect positive correlation
▪ 0 → no correlation
▪ -1 → perfect negative correlation
o Unlike covariance, it is unit-free and easier to interpret.

In this case:
• If the correlation is close to +1, it suggests that higher advertising
spend is strongly associated with higher daily sales.
Step 2: Python Code to Compute Correlation

```python
import numpy as np
# Given data
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]
# Convert lists to numpy arrays
x = np.array(advertising_spend)
y = np.array(daily_sales)
# Compute covariance
cov_xy = np.cov(x, y)[0, 1]
# Compute correlation coefficient
corr_xy = np.corrcoef(x, y)[0, 1]
# Print results
print("Covariance between advertising spend and daily sales:", cov_xy)
print("Correlation coefficient between advertising spend and daily sales:", corr_xy)
```

Interpretation
• Covariance = 28750 → Positive, indicating as advertising spend
increases, daily sales tend to increase.
• Correlation ≈ 0.992 → Very strong positive linear relationship
between advertising spend and daily sales.

Conclusion: Increasing advertising spend is strongly associated with
increased sales, so the marketing team’s spend appears effective.


Question 9: Your team has collected customer satisfaction survey data
on a scale of 1-10 and wants to understand its distribution before
launching a new product.  
● Explain which summary statistics and visualizations (e.g. mean,
standard deviation, histogram) you’d use.
● Write Python code to create a histogram using Matplotlib for the
survey data:  
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

Step 1: Understanding the Distribution
Before launching a new product, you want to understand how
customers rated satisfaction. For this, you can use:
Summary Statistics
1. Mean → Average satisfaction score; shows general customer
satisfaction.
2. Median → Middle value; helps if data is skewed.
3. Mode → Most frequent score; shows the most common rating.
4. Standard deviation (SD) → Measures how spread out the scores
are; low SD → most ratings are near the mean, high SD → ratings
vary widely.
Visualizations
1. Histogram → Shows the frequency of each score; easy to see
which scores are most common.
2. Boxplot → Shows median, quartiles, and potential outliers.
Step 2: Python Code to Create a Histogram


```python
import matplotlib.pyplot as plt
# Given survey scores
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]
# Calculate summary statistics
mean_score = sum(survey_scores) / len(survey_scores)
median_score = sorted(survey_scores)[len(survey_scores)//2]
mode_score = max(set(survey_scores), key=survey_scores.count)
std_dev = (sum((x - mean_score)**2 for x in survey_scores) /
len(survey_scores))**0.5
print("Mean:", mean_score)
print("Median:", median_score)
print("Mode:", mode_score)
print("Standard Deviation:", round(std_dev, 2))
# Create a histogram
plt.hist(survey_scores, bins=range(4, 12), edgecolor='black', alpha=0.7)
plt.title("Customer Satisfaction Survey Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.xticks(range(4, 11))
plt.show()
```

Explanation of the Code
• bins=range(4,12) → Creates bins for scores 4–10.
• edgecolor='black' → Adds borders for better visibility.
• alpha=0.7 → Slight transparency for aesthetics.
Interpretation
• Histogram shows which scores are most common.
• Summary statistics tell you:
o Mean ~ 7.5 → overall satisfaction is fairly high
o Median and mode show the most typical scores
o Standard deviation shows how varied the responses are

This helps your team decide whether the product is well-received
or needs improvement before launch.
